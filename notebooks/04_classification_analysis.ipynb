```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean = pd.read_csv('../data/processed/dataset_clean.csv')

target_classification = 'Addiction_Level'

plt.figure(figsize=(8, 5))
sns.countplot(x=target_classification, data=df_clean, palette='Set2')
plt.title(f'Distribuição das Classes: {target_classification}')
plt.xlabel('Nível de Vício (Classes Codificadas)')
plt.ylabel('Quantidade de Usuários')
plt.tight_layout()
plt.savefig('../reports/figures/class_balance.png')
plt.show()

feature_analise = 'Daily_Usage_Hours'

plt.figure(figsize=(8, 6))
sns.boxplot(x=target_classification, y=feature_analise, data=df_clean, palette='coolwarm')
plt.title(f'Relação: {target_classification} vs {feature_analise}')
plt.xlabel('Nível de Vício (Target)')
plt.ylabel('Horas de Uso Diário (Padronizado)')
plt.tight_layout()
plt.savefig('../reports/figures/boxplot_target_vs_feature.png')
plt.show()

df_clean.to_csv('../data/processed/dataset_classification.csv', index=False)
print("Dataset de Classificação salvo em: data/processed/dataset_classification.csv")
```
