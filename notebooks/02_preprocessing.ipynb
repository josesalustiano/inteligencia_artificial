```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder

df = pd.read_csv('../data/raw/dataset.csv')
print(f"Dimensões originais: {df.shape}")

colunas_numericas = df.select_dtypes(include=[np.number]).columns
colunas_categoricas = df.select_dtypes(exclude=[np.number]).columns

for col in colunas_numericas:
    df[col] = df[col].fillna(df[col].median())

for col in colunas_categoricas:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nValores nulos após imputação:")
print(df.isnull().sum().sum(), "nulos restantes.")

label_encoders = {}
for col in colunas_categoricas:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

print(f"\nDimensões após codificação: {df.shape}")

colunas_para_padronizar = [col for col in colunas_numericas if col != 'User_ID']

scaler = StandardScaler()
df[colunas_para_padronizar] = scaler.fit_transform(df[colunas_para_padronizar])


df.to_csv('../data/processed/dataset_clean.csv', index=False)
print("\nPré-processamento concluído! Arquivo 'dataset_clean.csv' salvo com sucesso em data/processed/.")
display(df.head())
```
