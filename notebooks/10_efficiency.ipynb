```python
import time
import tracemalloc
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

train_df = pd.read_csv('../data/processed/train_regression.csv')
test_df = pd.read_csv('../data/processed/test_regression.csv')
target_col = train_df.columns[-1]
X_train, y_train = train_df.drop(columns=[target_col]), train_df[target_col]
X_test = test_df.drop(columns=[target_col])

dt_model = DecisionTreeRegressor(random_state=42)
dt_model.fit(X_train, y_train)

def medir_eficiencia(modelo_predict, X, n_rep=30):
    tempos = []
    tracemalloc.start()
    
    for _ in range(n_rep):
        inicio = time.perf_counter()
        modelo_predict(X)
        tempos.append(time.perf_counter() - inicio)
        
    pico = tracemalloc.get_traced_memory()[1]
    tracemalloc.stop()
    
    tempo_medio = np.mean(tempos)
    energia_kwh = (65 * tempo_medio / 3600) / 1000 
    emissao_co2 = energia_kwh * 45.8 
    
    return {
        "Tempo Médio (s)": tempo_medio,
        "Vazão (amostras/s)": len(X) / tempo_medio,
        "Pico RAM (KB)": pico / 1024,
        "Energia (kWh)": energia_kwh,
        "Emissão CO2 (g)": emissao_co2
    }

resultados_eficiencia = medir_eficiencia(dt_model.predict, X_test)
print("Resultados de Eficiência Computacional:")
for k, v in resultados_eficiencia.items():
    print(f"{k}: {v:.6f}")
```
