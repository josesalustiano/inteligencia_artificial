```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score

# 1. Carregar os dados isolados
train_df = pd.read_csv('../data/processed/train_classification.csv')
test_df = pd.read_csv('../data/processed/test_classification.csv')
X_train, y_train = train_df.drop(columns=['Addiction_Level']), train_df['Addiction_Level']
X_test, y_test = test_df.drop(columns=['Addiction_Level']), test_df['Addiction_Level']

# 2. Treinamento: Random Forest
rf_clf = RandomForestClassifier(random_state=42, n_jobs=-1)
rf_clf.fit(X_train, y_train)
acc_rf = accuracy_score(y_test, rf_clf.predict(X_test))

# 3. Treinamento: Gradient Boosting
gb_clf = GradientBoostingClassifier(random_state=42)
gb_clf.fit(X_train, y_train)
acc_gb = accuracy_score(y_test, gb_clf.predict(X_test))

print(f"Acurácia Random Forest: {acc_rf * 100:.2f}%")
print(f"Acurácia Gradient Boosting: {acc_gb * 100:.2f}%")
```
