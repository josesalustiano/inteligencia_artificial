```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/dataset.csv')

display("Primeiras linhas do dataset:")
display(df.head())

display("Últimas linhas do dataset:")
display(df.tail())

print(f"\nDimensões do dataset: {df.shape[0]} linhas e {df.shape[1]} colunas")

print("\nTipos de dados de cada coluna:")
print(df.dtypes)

display("\nResumo estatístico das variáveis numéricas:")
display(df.describe())
```

```python
print("\n--- Valores Ausentes ---")
missing_data = df.isnull().sum()
missing_perc = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Qtd Ausentes': missing_data, 'Porcentagem (%)': missing_perc})
display(missing_df[missing_df['Qtd Ausentes'] > 0])


plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
plt.title('Mapa de Calor - Valores Ausentes')
plt.tight_layout()

plt.savefig('../reports/figures/missing_values.png')
plt.show()


print("\n--- Detecção de Outliers ---")

colunas_numericas = df.select_dtypes(include=[np.number]).columns

plt.figure(figsize=(12, 6))
sns.boxplot(data=df[colunas_numericas])
plt.xticks(rotation=45)
plt.title('Boxplot das Variáveis Numéricas (Identificação de Outliers)')
plt.tight_layout()

plt.savefig('../reports/figures/outliers_boxplot.png')
plt.show()
```

```python
# 7. Análise de Correlação
print("\n--- Matriz de Correlação ---")
# Selecionando apenas colunas numéricas (importante para evitar erros na correlação)
colunas_numericas = df.select_dtypes(include=[np.number])
matriz_correlacao = colunas_numericas.corr()

plt.figure(figsize=(14, 10))
# Criando o mapa de calor da correlação
sns.heatmap(matriz_correlacao, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Matriz de Correlação das Variáveis Numéricas')
plt.tight_layout()

plt.savefig('../reports/figures/correlation_matrix.png')
plt.show()

print("\n--- Visualizações de Distribuição e Relações ---")

plt.figure(figsize=(8, 5))
sns.histplot(df['Mental_Wellbeing_Score'].dropna(), kde=True, color='skyblue')
plt.title('Distribuição da Pontuação de Bem-Estar Mental')
plt.xlabel('Bem-Estar Mental (Score)')
plt.ylabel('Frequência')
plt.tight_layout()
plt.savefig('../reports/figures/hist_mental_wellbeing.png')
plt.show()


plt.figure(figsize=(8, 5))
sns.scatterplot(x='Daily_Usage_Hours', y='Mental_Wellbeing_Score', data=df, alpha=0.6, color='coral')
plt.title('Relação: Horas de Uso Diário vs Bem-Estar Mental')
plt.xlabel('Horas de Uso Diário')
plt.ylabel('Bem-Estar Mental (Score)')
plt.tight_layout()
plt.savefig('../reports/figures/scatter_usage_vs_wellbeing.png')
plt.show()


plt.figure(figsize=(10, 5))

sns.countplot(x='Primary_Platform', data=df, palette='Set2')
plt.title('Contagem de Usuários por Plataforma Principal')
plt.xlabel('Plataforma')
plt.ylabel('Quantidade de Usuários')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../reports/figures/bar_primary_platform.png')
plt.show()
```
